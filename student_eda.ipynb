```python
df['sex']=df['sex'].astype('category')
df['Pstatus']=df['Pstatus'].astype('category')
df['traveltime']=pd.to_datetime(df['traveltime'])
df['studytime']=pd.to_datetime(df['studytime'])
df['schoolsup']=df['schoolsup'].astype('category')
df['famsup']=df['famsup'].astype('category')
df['paid']=df['paid'].astype('category')
df['famsize']=df['famsize'].astype('category')
```

**Q1:Does study time affect grades?**

```python
df['studytime']
```

```python
df['studytime'].max()
```

```python
max_grade=df['G3'].max()
```

```python
df.loc[df['G3']== max_grade, 'studytime']
```

```python
sns.barplot(x='studytime', y='G3' , data=df, estimator='mean')
plt.xlabel('Study time')
plt.ylabel('Final grade')
plt.title('Study time VS Grades')
plt.show()
```

**The analysis shows a positive relationship between study time and final grades. Students who spend more time studying generally achieve higher average scores, although the improvement is not strictly linear, suggesting that factors beyond study time also influence performance.**

**Gender vs performance**

```python
sns.barplot(x='sex', y='G3', data=df)
plt.xlabel=('Gender')
plt.ylabel=('Grades')
plt.title=('Gender VS Performance')
plt.show()
```

**The bar plot indicates that male students have a slightly higher average final grade than female students, but the difference is minimal, suggesting that gender has a limited influence on academic performance.**
